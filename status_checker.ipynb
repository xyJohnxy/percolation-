# 2D scatter plot


```python
import matplotlib.pyplot as plt
import numpy as np

# Load the file
data = np.loadtxt("status_checker,N=100.dat")

x = data[:, 0]
y = data[:, 1]
z = data[:, 2]
status = data[:, 3]

# Create boolean masks for status filtering
unoccupied = status == 0
occupied = (status == 1) | (status == 2)
travelling = status == 3

fig = plt.figure(figsize=(10, 7))
# ax = fig.add_subplot(projection="3d")
ax = fig.add_subplot()

# Plot unoccupied points (Green, lower opacity)
ax.scatter(
    x[unoccupied],
    y[unoccupied],
    # z[unoccupied],
    c="green",
    alpha=0.05,
    marker="o",
    label="Unoccupied",
)

# Plot occupied & travelling points (Red, higher opacity)
ax.scatter(
    x[occupied],
    y[occupied],
    # z[occupied],
    c="red",
    alpha=0.8,
    marker="o",
    label="Occupied / Travelling",
)


ax.scatter(
    x[travelling],
    y[travelling],
    # z[occupied],
    c="blue",
    alpha=0.10,
    marker="o",
    label="Targets",
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
# ax.set_zlabel("Z")
ax.legend()
plt.show()
```

# 3D scatter plot

```python
import matplotlib.pyplot as plt
import numpy as np

# Load the file
data = np.loadtxt("3D_status_checker,N_pt=5000.dat")

x = data[:, 0]
y = data[:, 1]
z = data[:, 2]
status = data[:, 3]

# Create boolean masks for status filtering
unoccupied = status == 0
occupied = status == 1
travelling = status == 2
target = status == 3
retired = status == 4 

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(projection="3d")

# Plot unoccupied points (Green, lower opacity)
ax.scatter(
    x[unoccupied],
    y[unoccupied],
    z[unoccupied],
    c="green",
    alpha=1.0,
    marker="o",
    label="Unoccupied",
)

# Plot occupied & travelling points (Red, higher opacity)
ax.scatter(
    x[occupied],
    y[occupied],
    z[occupied],
    c="red",
    alpha=0.8,
    marker="o",
    label="Occupied",
)


ax.scatter(
    x[travelling],
    y[travelling],
    z[travelling],
    c="blue",
    alpha=1.0,
    marker="o",
    label="Travelling",
)
ax.scatter(
    x[target],
    y[target],
    z[target],
    c="yellow",
    alpha=1.0,
    marker="o",
    label="Targets",
)

ax.scatter(
    x[retired],
    y[retired],
    z[retired],
    c="gray",
    alpha=0.50,
    marker="o",
    label="Retired",
)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.legend()
plt.show()
```

# Interactive 3d version 

```python
import numpy as np
import plotly.graph_objects as go

# Load data
data = np.loadtxt("3D_status_checker,N_pt=5000.dat")

x = data[:, 0]
y = data[:, 1]
z = data[:, 2]
status = data[:, 3]

# Configuration mapping: (Label, Color, Opacity)
status_config = {
    0: ("Unoccupied", "green", 1),
    1: ("Occupied", "red", 0.80),
    2: ("Travelling", "blue", 1.00),
    3: ("Targets", "yellow", 1.00),
    4: ("Retired", "gray", 0.20),
}

fig = go.Figure()

# Add a trace for each status group
for code, (name, color, opacity) in status_config.items():
    mask = status == code
    fig.add_trace(
        go.Scatter3d(
            x=x[mask],
            y=y[mask],
            z=z[mask],
            mode="markers",
            name=name,
            marker=dict(
                size=3,
                color=color,
                opacity=opacity,
            ),
        )
    )

fig.update_layout(
    scene=dict(
        xaxis_title="X",
        yaxis_title="Y",
        zaxis_title="Z",
    ),
    width=900,
    height=700,
)

fig.show()
```
